# 개념 정리

## 4. 딥러닝 시작

### 4.1. 인공 신경망의 한계와 딥러닝 출현

퍼셉트론: 신경망의 기원이 되는 알고리즘. 다수의 신호를 입력으로 받아 하나의 신호를 출력하고, 이 신호를 입력으로 받아 1 또는 0 정보를 앞으로 전달.
- AND 게이트: 모든 입력이 1일 때 작동
- OR 게이트: 모든 입력이 0을 갖는 경우를 제외하고 작동
- XOR 게이트: 한개만 1일 때 작동. 비선형적으로 분리되어 단층 퍼셉트론으로 학습 불가능. 입력층과 출력층 사이에 중간층(은닉층)을 두어 다층 퍼셉트론을 구성하여 분류.
- 심층 신경망(딥러닝, Deep Neural Network): 입력층과 출력층 사이에 은닉층이 여러 개 있는 신경망

### 4.2. 딥러닝 구조

#### 4.2.1. 딥러닝 용어

신경망 구성 요소
- 층
    - 입력층(input layer): 입력
    - 은닉층(hidden layer): 모든 입력 노드로부터 입력 값을 받아 가중값을 계산하고 활성화 함수에 적용하여 출력층에 전달
    - 출력층(output layer): 최종 결괏값 포함
- 가중치(weight): 노드와 노드 간 연결 강도
- 바이어스(bias): 가중합에 더해주는 상수. 하나의 뉴런에서 활성화 함수를 거쳐 최종적으로 출력되는 값을 조절하는 역할
- 가중합(weighted sum), 전달 함수(transfer function): 가중치와 신호의 곱을 합한 가중합 계산 후 활성화 함수로 보냄
- 함수
    - 활성화 함수(activaton function): 신호를 입력받아 적절히 처리해 출력
    - 손실(loss function): 가중치 학습을 위해 출력 함수의 결과와 실제 값 간의 오차 측정

활성화 함수
- 시그모이드(sigmoid): 선형 함수 결과를 0~1 사이 비선형 형태로 변형. 주로 로지스틱 회귀와 같은 분류 문제를 확률적으로 표현할 때 사용. 딥러닝 모델의 깊이가 깊어지면 기울기가 사라지는 기울기 소멸 문제가 발생하는 문제가 있음.
- 하이퍼볼릭 탄젠트(hyperbolic tangent): 선형 함수 결과를 -1~1 사이 비선형 형태로 변형. 시그모이드와 달리 결괏값 평균을 0으로 만들었지만, 기울기 소멸 문제 발생.
- 렐루(ReLU): 입력이 음수일 때는 0을 출력하고, 양수일 때는 x를 출력. 경사 하강법에 영향을 주지 않아 학습 속도가 빠르고 기울기 소멸 문제가 발생하지 않는 장점. 일반적으로 은닉층에서 사용. 음수 값은 항상 0을 출력해 학습 능력이 감소한느 단점.
- 리키 렐루(Leaky ReLU): 렐루 함수의 단점을 보완해 입력 값이 음수이면 0이 아닌 매우 작은 수 반환.
- 소프트맥스(softmax): 입력 값을 0~1 사이에 출력되도록 정규화해 총합이 항상 1이 되도록 함. 출력 노드의 활성화 함수로 많이 사용.

손실 함수
- 평균 제곱 오차(MSE): 실제 값과 예측 값의 차이를 제곱하여 평균 낸 것. 회귀에서 주로 사용.
- 크로스 엔트로피 오차(Cross Entropy Error, CEE): 분류에서 원-핫 인코딩했을 때 사용할 수 있는 오차 계산법. 시그모이드와 MSE를 결합하여 사용하면 기울기가 매끄럽지 못하고 학습 속도가 늦어지는 단점을 보완.

#### 4.2.2. 딥러닝 학습

- 순전파(feedforward): 네트워크에 훈련 데이터가 들어올 때 발생. 모든 뉴런이 이전 층의 뉴런에서 수신한 정보에 변환을 적용해 다음 층의 뉴런으로 전송하는 방식. 데이터가 모든 층을 통과하여 예측 값이 최종 층에 도달하면 손실 함수로 오차 추정.  
- 역전파(backpropagation): 손실이 계산된 후 정보가 역으로 전파됨.

#### 4.2.3. 딥러닝의 문제점과 해결 방안

- 과적합(over-fitting) 문제: 훈련 데이터를 과하게 학습하여 발생. 실제 데이터에 대한 오차가 증가하는 현상. 드롭아웃(dropout)으로 임의로 일부 노드를 학습에서 제외시켜 해결.
- 기울기 소멸 문제: 은닉층이 많은 신경망에서 주로 발생. 렐루 활성화 함수 사용으로 해결.
- 성능이 나빠지는 문제: 경사 하강법으로 손실 함수의 비용이 최소가 되는 지점을 찾을 때까지 기울기가 낮은 쪽으로 이동시키는 과정을 반복할 때 성능이 나빠짐. 확률적 경사 하강법과 미니 배치 경사 하강법으로 해결.

- 배치 경사 하강법(Batch Gradient Descent, BGD): 전체 데이터셋에 대한 오류를 구한 후 기울기를 한 번만 계산하여 모델의 파라미터 업데이트. 전체 훈련 데이터셋에 대해 가중치 편미분. 학습이 오래걸리는 단점.
- 확률적 경사 하강법(Stochastic Gradient Descent, SGD): 임의로 선택한 데이터에 대해 기울기 계산. 빠른 계산이 가능하나 파라미터 변경 폭이 불안정하고 정확도가 낮을 수 있음.
- 미니 배치 경사 하강법(mini-batch gradient descent): 전체 데이터셋을 미니 배치 여러 개로 나누고, 미니 배치 한 개마다 기울기를 구한 후 평균 기울기를 이용해 모델을 업데이트. SGD보다 안정적이고 전체 데이터 학습보다 속도가 빠름.

옵티마이저로 학습 속도와 운동량 조정
- 속도 조정
    - 아다그라드(Adagrad, Adaptive gradient): 변수의 업데이트 횟수에 따라 학습률 조정. 많이 변화하지 않는 변수들의 학습률은 크게 하고, 많이 변화하는 학습률은 작게 조정. G함수로 파라미터마다 다른 학습률 부과. 기울기가 0에 수렴하는 문제가 있어 사용하지 않음.
    - 아다델타(Adadelta, Adaptive delta): 아다그라드에서 G값이 커짐에 다라 학습이 멈추는 문제를 해결. 학습률을 D함수로 변환하여 학습률에 대한 하이퍼파라미터가 필요하지 않음.
    - 알엠에스프롭(RMSProp): 아다그라드의 G값이 무한히 커지는 것을 방지. 감마를 추가하여 학습률 크기를 비율로 조정하도록 함.
- 운동량 조정
    - 모멘텀(Momentum): 가중치 수정 전 이전 수정 방향을 참고하여 같은 방향으로 일정한 비율만 수정. 지그재그 현상이 줄어들고, 관성 효과를 얻을 수 있는 장점이 있음. SGD와 함께 사용함.
    - 네스테로프 모멘텀(Nesterov Accelerated Gradient, NAG): 모멘텀 값이 적용된 지점에서 기울기 값 계산. 멈추어야 할 시점에서도 관성에 의해 훨씬 멀리 가는 모멘텀의 단점을 극복함.
- 속도와 운동량 혼용
    - 아담(Adam, Adaptive Moment Estimation): 모멘텀과 알엠에스프롭의 장점을 결합한 경사 하강법.

#### 4.2.4. 딥러닝을 사용할 때 이점

- 특성 추출(feature extraction): 데이터별로 어떤 특징을 가지고 있는지 찾아내 그것을 토대로 데이터를 벡터로 변환하는 작업. 딥러닝에서는 특성 추출 과정을 알고리즘에 통합시킴.
- 빅데이터의 효율적 활용: 데이터 사례가 많을수록 성능이 향상됨.

### 4.3. 딥러닝 알고리즘

#### 4.3.1. 심층 신경망

심층 신경망(DNN): 입력층과 출력층 사이에 다수의 은닉층을 포함하는 인공 신경망. 다양한 비선형적 관계를 학습할 수 있는 장점이 있지만 학습을 위한 연산량이 많고 기울기 소멸 문제 등이 발생할 수 있음.

#### 4.3.2. 합성곱 신경망

합성곱 신경망(Convolutional Nerual Network, CNN): 합성곱층(convolutinal layer)과 풀링층(pooling layer)을 포함하는 이미지 처리 성능이 좋은 인공 신경망 알고리즘. 대표적으로 LeNet-5와 AlexNet이 있음. 층을 더 깊게 쌓은 신경망으로 VGG, GoogLeNet, ResNet 등이 있음.

#### 4.3.3. 순환 신경망

순환 신경망(Recurrent Neural Network, RNN): 시계열 데이터 같은 시간 흐름에 따라 변화하는 데이터를 학습하기 위한 인공 신경망. 시간성(temporal property)을 가진 데이터가 많음. 데이터는 동적이고 길이가 가변적임. 기울기 소멸 문제가 있으며 LSTM으로 해결하고자 함. 자연어 처리 분야에서 주로 쓰임.

#### 4.3.4. 제한된 볼츠만 머신

제한된 볼츠만 머신(Restricted Boltzmann Machine, RBM): 가시층(visible layer)과 은닉층(hidden layer)로 구성되어, 가시층은 은닉층과만 연결되는 모델. 차원 감소, 분류, 선형 회귀 분석, 협업 필터링, 특성 값 학습, 주제 모델링에 사용. 기울기 소멸 문제를 해결하기 위해 사전 학습 용도로 활용 가능. 심층 신뢰 신경망(DBN)의 요소로 활용됨.

#### 4.3.5. 심층 신뢰 신경망

심층 신뢰 신경망(Deep Belief Network, DBN): 제한된 볼츠만 머신을 블록처럼 여러 층으로 쌓은 형태로 연결된 신경망. 레이블이 없는 데이터에 대한 비지도 학습 가능.순차적으로 학습시켜가면서 계층적 구조 생성. 위로 올라갈수록 추상적 특성을 추출함.

```python
import torch
```

```python
class Net(torch.nn.Module):
    def __init__(self, n_feature, n_hidden, n_output):
        super(Net, self).__init__()
        self.hidden = torch.nn.Linear(n_feature, n_hidden) # 은닉층
        self.relu = torch.nn.ReLu(inplace=True)
        self.out = torch.nn.Linear(n_hidden, n_output) # 출력층
        self.softmax = torch.nn.Softmax(dim=n_output)
    def forward(self, x):
        x = self.hidden(x)
        x = self.relu(x) # 은닉층을 위한 렐루 활성화 함수
        x = self.out(x)
        x = self.softmax(x) # 출력층을 위한 소프트맥스 활성화 함수
        return x
```

```python
import torch.nn.functional as F

class DropoutModel(torch.nn.Module):
    def  __init__(self):
        super(DropoutModel, self).__init__()
        self.layer1 = torch.nn.Linear(784, 1200)
        self.dropout1 = torch.nn.Dropout(0.5) # 50%의 노드를 무작위로 선택해 사용하지 않겠다는 의미
        self.layer2 = torch.nn.Linear(1200, 1200)
        self.dropout2 = torch.nn.Dropout(0.5)
        self.layer3 = torch.nn.Linear(1200, 10)

    def forward(self, x):
        x = F.relu(self.layer1(x))
        x = self.dropout1(x)
        x = F.relu(self.layer2(x))
        x = self.dropout2(x)
        return self.layer3(x)
```
