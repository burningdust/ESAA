# 개념 정리

## 5. 합성곱 신경망 I

### 5.1. 합성곱 신경망

#### 5.1.1. 합성곱층의 필요성
이미지 분석은 이미지를 펼쳐서 분석하여 데이터의 공간적 구조를 무시하게 되므로 합성곱층 도입

#### 5.1.2. 합성곱 신경망 구조
합성곱 신경망(CNN): 음성 인식이나 이미지/영상 인식에서 주로 사용되는 신경망.
- 입력층: 이미지 데이터가 최초로 거치는 계층. 이미지는 높이, 너비, 채널의 값을 갖는 3차원 데이터.
- 합성곱층(convolutional layer): 입력 데이터에서 특성을 추출하는 계층. 커널/필터가 이미지의 모든 영역을 훑으면서 특성 맵을 추출함. 커널은 3x3, 5x5 크기가 일반적. 스트라이드(stride)라는 지정된 간격에 따라 순차적으로 이동함. 필터 크기 안의 숫자 가중합을 계산한 후 스트라이드만큼 이동. 필터 채널이 3개인 컬러 이미지의 경우 각각 서로 다른 가중치로 합성곱을 적용해 더함.
- 풀링층(pooling layer): 특성 맵의 차원을 다운 샘플링해 연산량을 감소시키고 주요한 특성 벡터를 추출하는 계층.
    - 최대 풀링(max pooling): 대상 영역에서 최댓값을 추출. 대부분의 합성곱 신경망에서 사용.
    - 평균 풀링(average pooling): 대상 영역에서 평균을 반환.
- 완전연결층(fully connected layer): 차원이 축소된 특성맵을 전달받아 1차원 벡터로 펼침.
- 출력층: 소프트맥스 확성화 함수를 사용해 이미지가 각 레이블에 속할 확률 값을 출력하며, 가장 높은 확률 값을 갖는 레이블이 최종 값으로 선정됨.

#### 5.1.3. 1D, 2D, 3D 합성곱
- 1D 합성곱: 필터가 시간을 축으로 좌우로만 이동할 수 있는 합성곱. 그래프 곡선을 완화할 때 많이 사용됨. 입력(W), 필터(k) -> 출력(W)
- 2D 합성곱: 방향 두개로 움직이는 형태. 입력(W, H), 필터(k, k) -> 출력 (W, H)
- 3D 합성곱: 입력(W, H, L), 필터(k, k, d) -> 출력(W, H, L). d < L 유지가 중요.
- 3D 입력을 갖는 2D 합성곱: 입력(W, H, L), 필터(k, k, L) -> 출력(W, H). 대표적으로 LeNet-5와 VGG.
- 1x1 합성곱: 입력(W, H, L), 필터(1, 1, L) -> 출력(W, H). 채널 수를 조정해서 연산량이 감소되는 효과. 대표적으로 GoogLeNet.
